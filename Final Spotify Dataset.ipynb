```python
import pandas as pd

# Load the dataset
file_path = r"C:\Users\HP\Downloads\Spotify Dashboard\cleaned_data.csv"
df = pd.read_csv(file_path, encoding="latin-1")

# Replace blank spaces and strings like " " or "" with NaN
df.replace(r"^\s*$", pd.NA, regex=True, inplace=True)

# Replace zeros with NaN
df.replace(0, pd.NA, inplace=True)

# Drop rows where any column is NaN
df.dropna(inplace=True)

# Ensure numerical columns are clean (remove non-numeric characters and convert to numeric)
numerical_columns = ['Spotify Streams', 'YouTube Views', 'TikTok Views', 'Soundcloud Streams', 'TIDAL Popularity']  # Add relevant columns here
for col in numerical_columns:
    if col in df.columns:
        df[col] = df[col].replace({',': ''}, regex=True)  # Remove commas if present
        df[col] = pd.to_numeric(df[col], errors='coerce')  # Convert to numeric, coercing errors to NaN
        df.dropna(subset=[col], inplace=True)  # Drop rows with NaN in these columns after conversion

# Save the cleaned dataset
cleaned_file_path = r"C:\Users\HP\Downloads\Spotify Dashboard\fully_cleaned_data.csv"
df.to_csv(cleaned_file_path, index=False)

print(f"Cleaned data saved to {cleaned_file_path}")
```

```python
import pandas as pd

# Load the dataset
file_path = r"C:\Users\HP\Downloads\Spotify Dashboard\cleaned_data.csv"
df = pd.read_csv(file_path, encoding="latin-1")

# Strip leading/trailing spaces from column names
df.columns = df.columns.str.strip()

# Replace blank spaces, empty strings, '0' strings, and non-numeric values with NaN
df_cleaned = df.replace(r'^\s*$', pd.NA, regex=True)  # Replace blank strings
df_cleaned = df_cleaned.replace('0', pd.NA)          # Replace string '0'
df_cleaned = df_cleaned.replace(0, pd.NA)            # Replace numeric zero

# Apply the transformation using map for each column
df_cleaned = df_cleaned.apply(lambda col: col.map(lambda x: pd.NA if isinstance(x, str) and x.strip() == '' else x))

# Check missing data before dropping
print(df_cleaned.isna().sum())

# Drop rows with NaN values only for specific columns, or apply other strategies
df_cleaned = df_cleaned.dropna(subset=['Track', 'Artist'])

# Print results
print("Cleaned dataset shape:", df_cleaned.shape)
print(df_cleaned.head())

# Save cleaned dataset
output_path = r"C:\Users\HP\Downloads\Spotify Dashboard\fully_cleaned_data.csv"
df_cleaned.to_csv(output_path, index=False)
print("Cleaned dataset saved to:", output_path)
```
